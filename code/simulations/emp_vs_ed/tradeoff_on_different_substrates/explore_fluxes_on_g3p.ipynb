```python
from pathlib import Path

import cobra
import pandas as pd
import json
```

```python
model_path = "/Users/helenscott/Documents/PhD/Segre-lab/GEM-repos/GEM-mit1002/model/MIT1002-GEM.xml"
```

```python
# Load the model
model = cobra.io.read_sbml_model(model_path)
```

```python
minimal_media = {
    "EX_cpd00067_e0": 1000,  # H+_e0
    "EX_cpd00058_e0": 1000,  # Cu2+_e0
    "EX_cpd00007_e0": 20,  # O2_e0
    "EX_cpd00971_e0": 1000,  # Na+_e0
    "EX_cpd00063_e0": 1000,  # Ca2+_e0
    "EX_cpd00048_e0": 1000,  # Sulfate_e0
    "EX_cpd10516_e0": 1000,  # fe3_e0
    "EX_cpd00254_e0": 1000,  # Mg_e0
    "EX_cpd00009_e0": 1000,  # Phosphate_e0
    "EX_cpd00205_e0": 1000,  # K+_e0
    "EX_cpd00013_e0": 1000,  # NH3_e0
    "EX_cpd00099_e0": 1000,  # Cl-_e0
    "EX_cpd00030_e0": 1000,  # Mn2+_e0
    "EX_cpd00075_e0": 1000,  # Nitrite_e0
    "EX_cpd00001_e0": 1000,  # H2O_e0
    "EX_cpd00034_e0": 1000,  # Zn2+_e0
    "EX_cpd00149_e0": 1000,  # Co2+_e0
}
```

```python
gp3_medium = minimal_media.copy()
gp3_medium["EX_cpd00080_e0"] = 20  # Glycerol-3-phosphate_e0
model.medium = gp3_medium
```

```python
# Save reaction IDs for key reactions
# Define a reaction to indicate flux through the ED pathway
ed_rxn_id = "rxn01477_c0"
# Define a reaction to indicate flux through the EMP
emp_rxn_id = "rxn00558_c0"
# Define the biomass reaction ID
biomass_rxn_id = "bio1_biomass"
```

```python
# Force flux through the ED pathway and save the growth rate
with model:
    # Set the lowerbound of the ED reaction to the uptake rate
    model.reactions.get_by_id(ed_rxn_id).lower_bound = 20
    # Run pFBA
    try:
        solution = cobra.flux_analysis.pfba(model)
        # Save the fluxes to a json file
        fluxes = solution.fluxes.to_dict()
        # Save the fluxes to a json file
        name = "gp3_ed_forced_fluxes.json"
        with open(name, "w") as f:
            json.dump(fluxes, f)
    except:
        print(f"Error occurred while running pFBA with ED forcing on {name}")
```
